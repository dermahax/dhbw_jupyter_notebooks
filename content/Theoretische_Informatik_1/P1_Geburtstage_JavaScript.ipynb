## 1 · Ein zufälliger Hörsaal

`hoersaal(n)` liefert die Geburtstage von $n$ Personen als Array — das ist die
Wertetabelle der Abbildung $g$. (Wir nehmen 365 gleich wahrscheinliche Tage an.)

```javascript
function hoersaal(n) {
  // Geburtstage von n zufaelligen Personen, Tage 1..365
  var tage = [];
  for (var i = 0; i < n; i++) {
    tage.push(1 + Math.floor(Math.random() * 365));
  }
  return tage;
}

console.log(JSON.stringify(hoersaal(10)));
```
