## 1 · Ein zufälliger Hörsaal

`hoersaal(n)` liefert die Geburtstage von $n$ Personen als Liste — das ist die
Wertetabelle der Abbildung $g$. (Wir nehmen 365 gleich wahrscheinliche Tage an.)

```python
import random

def hoersaal(n):
    "Geburtstage von n zufaelligen Personen, Tage 1..365."
    return [random.randint(1, 365) for _ in range(n)]

print(hoersaal(10))
```
